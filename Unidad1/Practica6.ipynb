```python
n = len(auto)
K = int(1 + math.log2(n))
print("Valor de n = ", n, "\nNúmero de intervalos =", K)
```

```python
plt.hist(auto.price, bins=K, alpha=0.2, color='b', edgecolor="black")
plt.xlabel('Precio')
plt.ylabel('Frecuencia')
plt.title("Precio de los autos")
plt.show()
```

##**Interpretacion de Datos**

La mayoria de los autos se concentran en el rango de 5000-10000 USD, con una frecuencia de casi 100 unidades. Con el aumento del precio la frecuencia disminuye de forma constante hasta los 30000 USD. Entre los 30000-45000 USD se observa una pequeña variacion de baja frecuencia, la cual corresponde a automoviles de gama alta.

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=auto, x="body-style", y="price", palette='Set2')

plt.title('Relacion entre Body Style y Price')
plt.xlabel('Body Style')
plt.ylabel('Price')
plt.show()
```

##**Interpretacion de Datos**

El diagrama de caja y bigotes muestra la distribucion del precio segun el 'body style':


*   Categorias de menor precio: Hatchback, Sedan y Wagon
*   Categoria de mayor precio: Convertable y Hardtop
*   En los estilos Sedan, Hatchback y Wagon se observan puntos fuera de los bigotes superiores, lo que indica la presencia de modelos de gama alta






```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=auto, x="drive-wheels", y="price", palette='Set2')

plt.title('Relacion entre Drive Wheels y Price')
plt.xlabel('Drive Wheels')
plt.ylabel('Price')
plt.show()
```

##**Interpretacion de Datos**

El diagrama de caja y bigotes analiza la relacion entre el tipe de traccion y el precio de los vehiculos:


*   RWD: Presenta el rango de precios mas altos y variado del dataset de autos.
*   FWD: Concentra la mayoria de sus precios en el rango mas bajo y accesible
*   4WD: Muestra una distribucion compacta en el segmento economico de gama media





```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=auto, x="engine-location", y="price", palette='Set2')

plt.title('Relacion entre Engine Location y Price')
plt.xlabel('Engine Location')
plt.ylabel('Price')
plt.show()
```

##**Interpretacion de Datos**

El diagrama de caja y bigotes analiza la relacion entre la ubicacion del motor y el precio de los vehiculos:


*   Front: Concentra la inmensa mayoria de los vehiculos dentro del segmento economico y de gama media. Aunque el limite de sus bigotes llega cerca de los 29000 USD, cuenta con multiples valores atipicos que alcanzan hasta 45000 USD
*   Rear: Corresponde exclusivamente a un segmento de alto costo. Su muestra ronda los 35000 USD y muestra una caja extremadamente estrecha en la parte superior del grafico
