```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
# Parámetros de la distribución normal
media = 70
desv_std = 4
total_estudiantes = 500

peso_inferior = 68
peso_superior = 72

x = np.linspace(media - 4 * desv_std, media + 4 * desv_std, 1000)
y = norm.pdf(x, media, desv_std)
```

```python
prob_mas_68kg = 1 - norm.cdf(peso_inferior, media, desv_std)
est_mas_68kg = total_estudiantes * prob_mas_68kg
print('Probabilidad de que un estudiante pese más de 68 kg:', prob_mas_68kg)
print('Número de estudiantes que pesan más de 68 kg:', est_mas_68kg)


plt.figure(figsize=(6, 4))
plt.plot(x, y, 'b', linewidth=2)
plt.fill_between(x, y, where=(x >= peso_inferior), color='skyblue', alpha=0.5)
plt.axvline(peso_inferior, color='r', linestyle='--', label='68 kg')
plt.title(f'Probabilidad de que un estudiante pese más de 68 kg\n'
          f'P = {prob_mas_68kg:.4f}  |  n ≈ {est_mas_68kg:.0f}')
plt.xlabel('Peso (Kg)')
plt.ylabel('Densidad de la Probabilidad')
plt.legend()
plt.grid(True)
plt.show()
```

```python
prob_menos_72kg = norm.cdf(peso_superior, media, desv_std)
est_menos_72kg = total_estudiantes * prob_menos_72kg
print('Probabilidad de que un estudiante pese menos de 72 kg:', prob_menos_72kg)
print('Número de estudiantes que pesan menos de 72 kg:', est_menos_72kg)

plt.figure(figsize=(6, 4))
plt.plot(x, y, 'b', linewidth=2)
plt.fill_between(x, y, where=(x <= peso_superior), color='lightgreen', alpha=0.5)
plt.axvline(peso_superior, color='g', linestyle='--', label='72 kg')
plt.title(f'Probabilidad de que un estudiante pese menos de 72 kg\n'
          f'P = {prob_menos_72kg:.4f}  |  n ≈ {est_menos_72kg:.0f}')
plt.xlabel('Peso (Kg)')
plt.ylabel('Densidad de la Probabilidad')
plt.legend()
plt.grid(True)
plt.show()
```

```python
prob_entre_68_72 = norm.cdf(peso_superior, media, desv_std) - norm.cdf(peso_inferior, media, desv_std)
est_entre_68_72 = total_estudiantes * prob_entre_68_72
print('Probabilidad de que un estudiante pese entre 68 y 72 kg:', prob_entre_68_72)
print('Número de estudiantes que pesan entre 68 y 72 kg:', est_entre_68_72)
print('Número aproximado (redondeado):', round(est_entre_68_72))

plt.figure(figsize=(6, 4))
plt.plot(x, y, 'b', linewidth=2)
plt.fill_between(x, y, where=((x >= peso_inferior) & (x <= peso_superior)),
                 color='orange', alpha=0.5)
plt.axvline(peso_inferior, color='r', linestyle='--', label='68 kg')
plt.axvline(peso_superior, color='g', linestyle='--', label='72 kg')
plt.title(f'Probabilidad de que un estudiante pese entre 68 y 72 kg\n'
          f'P = {prob_entre_68_72:.4f}  |  n ≈ {est_entre_68_72:.0f}')
plt.xlabel('Peso (Kg)')
plt.ylabel('Densidad de la Probabilidad')
plt.legend()
plt.grid(True)
plt.show()
```
