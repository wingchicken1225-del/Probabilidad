```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, poisson
```

```python
n = 1000      # Motores por lote
p = 0.001     # Probabilidad de que un motor sea defectuoso
lam = n * p   # Media para la aproximación de Poisson (λ = 1)

prob_0 = binom.pmf(0, n, p)
prob_3 = binom.pmf(3, n, p)
prob_4 = binom.pmf(4, n, p)

prob_0_p = poisson.pmf(0, lam)
prob_3_p = poisson.pmf(3, lam)
prob_4_p = poisson.pmf(4, lam)

print('Resultados con la distribución Binomial (n=1000, p=0.001):')
print('P(X = 0) ninguno defectuoso   :', prob_0)
print('P(X = 3) tres defectuosos     :', prob_3)
print('P(X = 4) cuatro defectuosos   :', prob_4)
print()
print('Resultados con la aproximación de Poisson (λ=1):')
print('P(X = 0):', prob_0_p)
print('P(X = 3):', prob_3_p)
print('P(X = 4):', prob_4_p)
```

```python
x = np.arange(0, 11)
y = binom.pmf(x, n, p)

colores = []
for k in x:
    if k == 0:
        colores.append('orange')
    elif k == 3:
        colores.append('lightgreen')
    elif k == 4:
        colores.append('red')
    else:
        colores.append('skyblue')

plt.figure(figsize=(7, 4))
plt.bar(x, y, color=colores, edgecolor='black')
plt.title('Distribución Binomial (n = 1000, p = 0.001)\n'
          f'P(0) = {prob_0:.4f} | P(3) = {prob_3:.4f} | P(4) = {prob_4:.4f}')
plt.xlabel('Número de motores defectuosos (X)')
plt.ylabel('Probabilidad')
plt.xticks(x)
plt.grid(True, axis='y')
plt.show()

ancho = 0.4
plt.figure(figsize=(7, 4))
plt.bar(x - ancho/2, binom.pmf(x, n, p), width=ancho, color='skyblue',
        edgecolor='black', label='Binomial')
plt.bar(x + ancho/2, poisson.pmf(x, lam), width=ancho, color='orange',
        edgecolor='black', label='Poisson (λ=1)')
plt.title('Comparación: Binomial vs Aproximación de Poisson')
plt.xlabel('Número de motores defectuosos (X)')
plt.ylabel('Probabilidad')
plt.xticks(x)
plt.legend()
plt.grid(True, axis='y')
plt.show()
```
