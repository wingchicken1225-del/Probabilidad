```python
from scipy.stats import poisson
```

```python
# Parametros para la distribucion de Poisson
media_muertes = 5000 * 0.001 # 5000 hombres con una probabilidad de muerte de 0.001

# Numero de reclamaciones
x = 4

#Calcular la probabilidad de que la campañia tenga que pagar 4 reclamaciones en un año determinado
probabilidad_4_reclamaciones = poisson.pmf(x, media_muertes)

print('La probabilidad de que la campañia tenga que pagar por 4 reclamaciones es DE: ', probabilidad_4_reclamaciones)
```
