```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson
```

```python
# Parámetros de la distribución de Poisson
mu = 5  # Media (personas por día)

prob_igual_2 = poisson.pmf(2, mu)

print('a) Probabilidad de que ingresen exactamente 2 personas en un día:', prob_igual_2)
print('   Porcentaje:', round(prob_igual_2 * 100, 2), '%')

prob_mas_10 = 1 - poisson.cdf(10, mu)   # equivalente a poisson.sf(10, mu)

x = np.arange(0, 16)
y = poisson.pmf(x, mu)

colores_a = ['orange' if k == 2 else 'skyblue' for k in x]

plt.figure(figsize=(6, 4))
plt.bar(x, y, color=colores_a, edgecolor='black')
plt.title(f'Distribución de Poisson (μ = {mu})\n'
          f'P(X = 2) = {prob_igual_2:.4f}')
plt.xlabel('Número de personas ingresadas (X)')
plt.ylabel('Probabilidad')
plt.xticks(x)
plt.grid(True, axis='y')
plt.show()
```

```python
print('b) Probabilidad de que ingresen más de 10 personas en un día:', prob_mas_10)
print('   Porcentaje:', round(prob_mas_10 * 100, 2), '%')

colores_b = ['red' if k > 10 else 'skyblue' for k in x]

plt.figure(figsize=(6, 4))
plt.bar(x, y, color=colores_b, edgecolor='black')
plt.axvline(10.5, color='k', linestyle='--', label='X > 10')
plt.title(f'Distribución de Poisson (μ = {mu})\n'
          f'P(X > 10) = {prob_mas_10:.4f}')
plt.xlabel('Número de personas ingresadas (X)')
plt.ylabel('Probabilidad')
plt.xticks(x)
plt.legend()
plt.grid(True, axis='y')
plt.show()
```

##Conclusion

En una distribución de Poisson, X puede tomar cualquier valor entero 0, 1, 2, 3, ... sin límite superior. Por eso, 11, 12, 15 o incluso 30 personas en un día son valores que, en teoría, pueden ocurrir. Su probabilidad nunca llega a ser exactamente cero. Es posible que X exceda de 10, pero es poco probable (≈ 1.37 %), porque 10 está muy por encima del promedio de 5 personas por día.
