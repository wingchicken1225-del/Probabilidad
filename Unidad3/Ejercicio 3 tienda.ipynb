```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x3 = [0, 1, 2, 3, 4, 5]
p3 = [0.05, 0.10, 0.25, 0.30, 0.20, 0.10]

suma_p3 = sum(p3)
print(f"Suma de probabilidades: {suma_p3:.2f}\n")

valor_esperado_3 = 0
for i in range(len(x3)):
    xi = x3[i]
    pi = p3[i]
    producto = xi * pi
    valor_esperado_3 += producto
    print(f"x = {xi}, p = {pi}, x*p = {producto:.2f}")

print(f"\nValor esperado E(X) (Ventas promedio): {valor_esperado_3:.4f}")

e_x2_3 = 0
for i in range(len(x3)):
    xi = x3[i]
    pi = p3[i]
    e_x2_3 += (xi ** 2) * pi

var3 = e_x2_3 - (valor_esperado_3 ** 2)
desv3 = np.sqrt(var3)
print(f"Varianza: {var3:.4f}")
print(f"Desviación Estándar: {desv3:.4f}\n")

f_acum3 = np.cumsum(p3)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.stem(x3, p3)
plt.title("Distribución de Probabilidad - Ej. 3")
plt.xlabel("Unidades vendidas (x)")
plt.ylabel("Probabilidad P(X)")
plt.grid(True)

plt.subplot(1, 2, 2)
plt.step(x3, f_acum3, where='mid', marker='o')
plt.title("Distribución Acumulada F(x) - Ej. 3")
plt.xlabel("Unidades vendidas (x)")
plt.ylabel("F(x)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

##Interpretacion de Datos



*   Cantidad de unidades a vender por semana ($E(X)$): El valor esperado indica el número promedio de artículos que la tienda vende semanalmente a largo plazo, sirviendo como base fundamental para la planeación del inventario.
*   Estabilidad de la demanda: La varianza y la desviación estándar permiten analizar si las ventas semanales son estables (poco dispersas alrededor del promedio) o si presentan fluctuaciones drásticas, lo que ayuda a la administración a anticipar riesgos de desabastecimiento o exceso de producto.
