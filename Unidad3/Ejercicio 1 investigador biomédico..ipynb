```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x1 = [0, 1, 2, 3, 4, 5]
p1 = [0.10, 0.25, 0.30, 0.20, 0.10, 0.05]

suma_p1 = sum(p1)
print(f"Suma de probabilidades: {suma_p1:.2f}\n")

valor_esperado_1 = 0
for i in range(len(x1)):
    xi = x1[i]
    pi = p1[i]
    producto = xi * pi
    valor_esperado_1 += producto
    print(f"x = {xi}, p = {pi}, x*p = {producto:.2f}")

print(f"\nValor esperado E(X): {valor_esperado_1:.4f}")

e_x2_1 = 0
for i in range(len(x1)):
    xi = x1[i]
    pi = p1[i]
    e_x2_1 += (xi ** 2) * pi

var1 = e_x2_1 - (valor_esperado_1 ** 2)
desv1 = np.sqrt(var1)
print(f"Varianza: {var1:.4f}")
print(f"Desviación Estándar: {desv1:.4f}\n")

f_acum1 = np.cumsum(p1)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.stem(x1, p1)
plt.title("Distribución de Probabilidad - Ej. 1")
plt.xlabel("Células anormales (x)")
plt.ylabel("Probabilidad P(X)")
plt.grid(True)

plt.subplot(1, 2, 2)
plt.step(x1, f_acum1, where='mid', marker='o')
plt.title("Distribución Acumulada F(x) - Ej. 1")
plt.xlabel("Células anormales (x)")
plt.ylabel("F(x)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

##Interpretacion de Datos



*   Promedio esperado de células anormales por muestra ($E(X)$): El valor esperado representa el número promedio de células anormales encontradas en una muestra seleccionada al azar de una lámina.
*   Variabilidad entre las muestras: La varianza y la desviación estándar miden qué tan dispersos están los conteos de células respecto al promedio. Una desviación estándar baja indica que la mayoría de las muestras se concentran cerca del valor esperado, mientras que un valor alto mostraría fluctuaciones considerables entre los tejidos de distintos pacientes
*   Función de distribución acumulada ($F(x)$): Indica la probabilidad de encontrar un número menor o igual a un determinado conteo de células anormales ($P(X \le x)$). Permite evaluar rápidamente qué tan probable es hallar muestras con baja o alta carga celular anormal.
