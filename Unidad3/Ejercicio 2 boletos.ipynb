```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x2 = [0, 10, 20, 50]
p2 = [0.50, 0.30, 0.15, 0.05]

suma_p2 = sum(p2)
print(f"Suma de probabilidades: {suma_p2:.2f}\n")

valor_esperado_2 = 0
for i in range(len(x2)):
    xi = x2[i]
    pi = p2[i]
    producto = xi * pi
    valor_esperado_2 += producto
    print(f"x = {xi}, p = {pi}, x*p = {producto:.2f}")

print(f"\nValor esperado E(X) (Ganancia promedio): {valor_esperado_2:.4f}")

e_x2_2 = 0
for i in range(len(x2)):
    xi = x2[i]
    pi = p2[i]
    e_x2_2 += (xi ** 2) * pi

var2 = e_x2_2 - (valor_esperado_2 ** 2)
desv2 = np.sqrt(var2)
print(f"Varianza: {var2:.4f}")
print(f"Desviación Estándar: {desv2:.4f}\n")

f_acum2 = np.cumsum(p2)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.stem(x2, p2)
plt.title("Distribución de Probabilidad - Ej. 2")
plt.xlabel("Monto en pesos (x)")
plt.ylabel("Probabilidad P(X)")
plt.grid(True)

plt.subplot(1, 2, 2)
plt.step(x2, f_acum2, where='mid', marker='o')
plt.title("Distribución Acumulada F(x) - Ej. 2")
plt.xlabel("Monto en pesos (x)")
plt.ylabel("F(x)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

##Interpretacion de Datos


*   Significado del valor esperado ($E(X)$): Representa la ganancia promedio esperada por boleto si el juego se repitiera un número grande de veces.
*   ¿El juego es favorable para el jugador?: Si el valor esperado de la ganancia es menor al costo pagado por el boleto, el juego no es favorable para el jugador (favorece a la casa o organizador). Si fuera igual a cero, se consideraría un juego justo.
